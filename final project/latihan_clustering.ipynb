16. Pivot Table for Order Status: Create a pivot table to summarize the count of orders (order_id) by order_status for each customer_state. Display customer_state as the rows, order_status as the columns, and the count of orders as the values.

```python
# Create a pivot table to summarize the count of orders by order_status for each customer_state
pivot_table = df.pivot_table(
    index='customer_state',  # Rows
    columns='order_status',  # Columns
    values='order_id',       # Values
    aggfunc='count',
    fill_value=0   
)

# Display the pivot table
pivot_table
```

17. Group by Cluster and Aggregate: Write a query to group the dataset by Cluster and calculate the total Monetary value and the average Recency for each cluster. Display Cluster, total Monetary, and average Recency.

```python
cluster = df.groupby('Cluster').agg({
    'Monetary':['sum'],
    'Recency':['mean']
}).reset_index()
cluster.columns = ['Cluster','Monetary Total','Average Recency']
cluster
```

```python
df.groupby('customer_city').agg(
    total_order = ('order_id','nunique'),
    average_monetary = ('Monetary','mean')
).reset_index()
```
