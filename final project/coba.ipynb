```python
import pandas as pd

# Convert 'order_purchase_timestamp' to datetime format if not already done
df['order_purchase_timestamp'] = pd.to_datetime(df['order_purchase_timestamp'])

# Filter data for the customer state 'SP' and date range in August 2018
product = df.loc[
    (df['customer_state'] == 'SP') & 
    (df['order_purchase_timestamp'] >= '2018-08-01') & 
    (df['order_purchase_timestamp'] <= '2018-08-31')
]

# Group by 'product_category_name_english' and aggregate relevant columns
product_sp = product.groupby('product_category_name_english').agg(
    order_count=('order_id', 'nunique'),
    first_purchase=('order_purchase_timestamp', 'min'),
    last_purchase=('order_purchase_timestamp', 'max'),
    unique_states=('customer_state', 'nunique')
).reset_index()

# Replace 'unique_states' value 1 with 'SP'
product_sp['unique_states'] = product_sp['unique_states'].replace(1, 'SP')

# Sort the results by 'order_count' in descending order
product_sp_sorted = product_sp.sort_values('order_count', ascending=False)

# Display the result
product_sp_sorted
```
