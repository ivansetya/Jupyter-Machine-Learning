```python
from scipy.stats import normaltest, skew

numerical_features = df.describe().columns
```

```python
column_colors = {
    'Duration': '#C39BD3',       
    'Net Sales': '#F7DC6F', 
    'Commision (in value)': '#7DCEA0',
    'Age': '#85C1E9'
}

for col in numerical_features:
    fig, (ax1, ax2) = plt.subplots(2, sharex=True, figsize=(7.5, 5), 
                                    gridspec_kw={"height_ratios": (.15, .85),
                                                "hspace": 0.05})

    # Test for normal distribution and calculate skewness
    stat, pvalue = normaltest(df[col].dropna())
    skewness = skew(df[col].dropna())

    if pvalue > 0.05:
        distribution_info = f'p = {pvalue:.3f} (Normal Distribution)\nSkewness = {skewness:.2f}'
        hist_color = column_colors.get(col, '#85C1E9')
        box_color = 'white'
    else:
        if skewness > 0:
            skew_type = 'Right Skewed'
        elif skewness < 0:
            skew_type = 'Left Skewed'
        else:
            skew_type = 'Symmetric'

        distribution_info = f'p = {pvalue:.3f} (NOT Normal Distribution)\nSkewness = {skewness:.2f} ({skew_type})'
        hist_color = column_colors.get(col, '#85C1E9')
        box_color = column_colors.get(col, '#85C1E9')

    ax1.set_title(f"Distribution of {col}\n{distribution_info}", fontsize=10, color="black")
    sns.boxplot(x=df[col], ax=ax1, color=box_color, linecolor='black', flierprops=dict(marker='d', markersize=6, markeredgecolor='None', markerfacecolor='black'))
    ax1.set(xlabel="")

    sns.histplot(data=df[col], ax=ax2, stat="probability", kde=True, bins=50, color=hist_color)

    plt.show()
```

> ### **Identify Outliers**

```python
def outlier_stats(column):
    Q1 = column.quantile(0.25)
    Q3 = column.quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    outliers = column[(column < lower_bound) | (column > upper_bound)]
    count_outliers = len(outliers)
    percentage_outliers = (count_outliers / len(column)) * 100
    return count_outliers, percentage_outliers

colors = ['#C39BD3', '#F7DC6F', '#7DCEA0', '#85C1E9']

fig, axes = plt.subplots(nrows=2, ncols=2, figsize=(15, 10))
axes = axes.flatten()

for i, col in enumerate(df.describe().columns):
    count_outliers, pct_outliers = outlier_stats(df[col])
    sns.boxplot(data=df[col], color=colors[i], ax=axes[i], orient='h', linecolor='black',
                flierprops=dict(marker='d', markersize=6, markeredgecolor='None', markerfacecolor='black'))
    axes[i].set_title(f"{col.replace('_', ' ').title()} \n{count_outliers} outliers ({pct_outliers:.2f}%)", fontsize=12)

plt.tight_layout()
plt.show()
```

1. **Single Trip Travel Insurance**
- **Coverage**: Covers a single trip from departure to return.
- **Covered Days**: Typically covers trips up to a few weeks or months, depending on the policy.
- **Advantages**: Tailored coverage for specific trip dates and activities.
- **Suitable for**: Occasional travelers or those taking one-time trips within a year.
2. **Annual Multi-Trip Travel Insurance**
- **Coverage**: Covers multiple trips within a year under one policy.
- **Covered Days**: Each trip usually limited to 30-60 days per trip (varies by insurer).
- **Advantages**: Convenient and cost-effective for frequent travelers.
- **Suitable for**: Travelers who take multiple short trips annually.
3. **90-Day Travel Insurance**
- **Coverage**: Specifically for trips up to 90 days.
- **Covered Days**: Covers trips lasting up to three months.
- **Advantages**: Cost-effective for longer individual trips.
- **Suitable for**: Backpackers, extended vacations, or long cruises.
4. **Long Stay (Backpacker) Travel Insurance**
- **Coverage**: Designed for trips lasting 3 months to 18 months.
- **Covered Days**: Covers extended stays abroad.
- **Advantages**: Comprehensive coverage for longer travel periods.
- **Suitable for**: Gap years, working holidays, or extended sabbaticals abroad.
