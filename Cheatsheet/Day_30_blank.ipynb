```python
# libraries
import pandas as pd
import numpy as np
import statsmodels.api as sm
from statsmodels.stats.outliers_influence import variance_inflation_factor
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.metrics import confusion_matrix
```

```python
# import dataset
df = pd.read_csv('../Dataset/bankloan.csv')
df
```

```python
# make X and y
x = sm.add_constant(df[['employ','debtinc','creddebt','othdebt']])
y = df['default']
display(x,y)
```

```python
# model fitting
sm_logit = sm.Logit(y,x)
result = sm_logit.fit()
```

```python
print(result.summary())
```

```python
def calc_vif(X):

    # Calculating VIF
    vif = pd.DataFrame()
    vif["variables"] = X.columns
    vif["VIF"] = [variance_inflation_factor(X.values, i) for i in range(X.shape[1])]

    return(vif)
```

```python
# collinearity check
calc_vif(df[['employ','debtinc','creddebt','othdebt']])
```

```python
 # Model interpretation
x.describe().loc[['min','max']].T
```

```python

# Parameter Estimate
# OR ~ Odd Ratio
# employment tenure
B1 = -0.2436
c = 25
d = 24
OR1 = np.exp(B1*(c-d))
OR1
# between 0 years - 31 years employment tenure, when employment tenure increase by 1 year and no changes in another variable,
# the risk to become default most likely increase by 0.7838010920039364 times.
```

```python
m = 1/OR1
m
# the risk to become default most likely decrease by 1.2758338948511925 times.
```

```python
# debt ratio against income
B2 = 0.0885
c = 0.42
d = 0.41
OR2 = np.exp(B2*(c-d))
OR2
# between 0.4 % - 41.3 % debt ratio to income, when the ratio increase by 1% and no change in another variable,
# the risk to become defult most likely increase by 1.0008853917280514 % times.
```

```python
# credit amount
B3 = 0.5041
c = 10
d = 5
OR3 = np.exp(B3*(c-d))
OR3
# between 0.01 $ - 20.56 $ credit amount, when the credit amount increase by 5 $ and no change in another variable,
# the risk to become defauult most likekly increase by 12.434812515742879 times.
```

LLR-TEST:

B1, B2, B3, B4: P-value of the log-likelyhood-ratio (2.473e-48) is below than 0.05. we can say we have enough evidence that either employ, debtinc, creditdebt, othdebt have significant effect to credit default rate.

WALD-TEST:
<br>
1. B0: P-value of the partial test is below than 0.05. means, B0 needed for the model
1. B1: P-value of the partial test is below than 0.05. means, we have enough evidence that employ decrease the default rate
1. B2: P-value of the partial test is below than 0.05. means, we have enough evidence that debtinc increase the default rate
1. B3: P-value of the partial test is below than 0.05. means, we have enough evidence that creditdebt increase the default rate
1. B4: P-value of the partial test is more than 0.05. means, we DON'T have enough evidence that othdebt has significant effect with default rate

COEFFICIENT DETERMINATION:

Pseudo R-Square: 28.44 %, this model can explain 28.44 $ variation of default rate.

```python
# SKLEARN ! -> feature selection
x_train, x_test, y_train, y_test = train_test_split(
x,
y,
stratify=y, # make even for 0 and 1 between train and test
test_size=0.3,
random_state=2024
)
```

```python
# check the target value proportion
display(y_train.value_counts(normalize=True),y_test.value_counts(normalize=True))
```

```python
# model fitting
sm_logit_train = sm.Logit(y_train,x_train)
result_train = sm_logit_train.fit()
```

```python
# predict using the model (result_train)
y_predict_proba = result_train.predict(x_test)
y_predict_proba
```

```python
# convert proba into class (0 and 1)
y_predict_class = np.where(y_predict_proba>0.5,1,0)
y_predict_class
```

```python
# accuracy
print('model accuracy:',accuracy_score(y_test,y_predict_class))
```

```python
# confusion_matrix
confusion_matrix(y_test,y_predict_class)
```
